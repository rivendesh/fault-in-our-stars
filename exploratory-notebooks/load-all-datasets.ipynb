> Aggregating data to stitch all quarters together

```python
KCC_processed = KCC.copy()
KCC_processed = KCC_processed.sort_values(by=['KIC', 'quarter'], ascending=[True, True]).reset_index(drop=True)
```

```python
KCC_processed.head(25)
```

```python
def concatenate_list_columns(group, col_name):
    all_data = group[col_name].tolist()
    
    # Filter and prepare data: only keep non-empty list/array items
    valid_data_for_concat = [item for item in all_data if isinstance(item, (list, np.ndarray)) and len(item) > 0]
    
    if valid_data_for_concat:
        # Convert list elements to arrays, then concatenate
        try:
            return np.concatenate([np.array(item) for item in valid_data_for_concat]).tolist()
        except Exception:
            # Fallback if concatenation fails unexpectedly
            return []
    return []
```

```python
def aggregate_kic_data(group):
    total_n_points = group['n_points'].sum()

    def concatenate_list_columns(group, col_name):
        all_data = group[col_name].tolist()
        
        # Filter and prepare data: only keep non-empty list/array items
        valid_data_for_concat = [item for item in all_data if isinstance(item, (list, np.ndarray)) and len(item) > 0]
        
        if valid_data_for_concat:
            # Convert list elements to arrays, then concatenate
            try:
                return np.concatenate([np.array(item) for item in valid_data_for_concat]).tolist()
            except Exception:
                # Fallback if concatenation fails unexpectedly
                return []
        return []

    merged_time = concatenate_list_columns(group, 'time')
    merged_detflux = concatenate_list_columns(group, 'detflux')
    merged_detflux_err = concatenate_list_columns(group, 'detflux_err')

    return pd.Series({
        'time': merged_time,
        'flux': merged_detflux,
        'flux_err': merged_detflux_err,
        'n_points': total_n_points
    })

KCC_processed = KCC.copy()

# Aggregate data by KIC
result_df = KCC_processed.groupby('KIC').apply(aggregate_kic_data)
```

> Plotting a LC from the stitched dataframe

```python
def plotLightCurve(df: pd.DataFrame, kic: int, with_error: bool = True):
    """
    Plots the light curve for a given KIC from a DataFrame.
    Assumes KIC is the index of the input DataFrame 'df'.
    """
    # 1. Retrieve the entire row using the index label 'kic'
    # This returns a Pandas Series object, which is what we want.
    try:
        row_series = df.loc[kic]
    except KeyError:
        raise ValueError(f"KIC {kic} not found in DataFrame index.")
    
    # Check if the retrieved row is empty (though KeyError should handle most missing cases)
    if row_series.empty:
        raise ValueError(f"KIC {kic} resulted in an empty record in the DataFrame.")

    # 2. Access data directly using string keys on the Series
    t = row_series["time"]
    f = row_series["flux"]
    e = row_series.get("flux_err", None)

    # 3. Plotting
    plt.figure(figsize=(20, 5))
    if with_error and e is not None:
        plt.errorbar(t, f, yerr=e, fmt='-', linewidth=1)
    else:
        plt.plot(t, f, linewidth=1)
        
    plt.xlabel("Time")
    plt.ylabel(r"Normalized Flux (e$^{-}$ s$^{-1}$)")
    plt.title(f"KIC {kic} Light Curve")
    plt.show()
```

```python
def saveLightCurve(df: pd.DataFrame, kic: int, with_error: bool = True):
    """
    Plots the light curve for a given KIC from a DataFrame.
    Assumes KIC is the index of the input DataFrame 'df'.
    """
    # 1. Retrieve the entire row using the index label 'kic'
    # This returns a Pandas Series object, which is what we want.
    try:
        row_series = df.loc[kic]
    except KeyError:
        raise ValueError(f"KIC {kic} not found in DataFrame index.")
    
    # Check if the retrieved row is empty (though KeyError should handle most missing cases)
    if row_series.empty:
        raise ValueError(f"KIC {kic} resulted in an empty record in the DataFrame.")

    # 2. Access data directly using string keys on the Series
    t = row_series["time"]
    f = row_series["flux"]
    e = row_series.get("flux_err", None)

    # 3. Plotting
    plt.figure(figsize=(20, 5))
    plt.plot(t, f, linewidth=1)
    
    plt.xlabel("Time")
    plt.ylabel(r"Normalized Flux (e$^{-}$ s$^{-1}$)")
    plt.title(f"KIC {kic} Light Curve")
    # plt.show()
    
    filename = f"light_curve_{kic}.png"
    plt.savefig(filename)

    plt.close()
```
