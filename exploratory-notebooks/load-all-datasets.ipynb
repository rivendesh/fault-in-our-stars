```python
def concatenate_list_columns(group, col_name):
    all_data = group[col_name].tolist()
    
    # Filter and prepare data: only keep non-empty list/array items
    valid_data_for_concat = [item for item in all_data if isinstance(item, (list, np.ndarray)) and len(item) > 0]
    
    if valid_data_for_concat:
        # Convert list elements to arrays, then concatenate
        try:
            return np.concatenate([np.array(item) for item in valid_data_for_concat]).tolist()
        except Exception:
            # Fallback if concatenation fails unexpectedly
            return []
    return []
```

```python
def aggregate_kic_data(group):
    total_n_points = group['n_points'].sum()

    def concatenate_list_columns(group, col_name):
        all_data = group[col_name].tolist()
        
        # Filter and prepare data: only keep non-empty list/array items
        valid_data_for_concat = [item for item in all_data if isinstance(item, (list, np.ndarray)) and len(item) > 0]
        
        if valid_data_for_concat:
            # Convert list elements to arrays, then concatenate
            try:
                return np.concatenate([np.array(item) for item in valid_data_for_concat]).tolist()
            except Exception:
                # Fallback if concatenation fails unexpectedly
                return []
        return []

    merged_time = concatenate_list_columns(group, 'time')
    merged_detflux = concatenate_list_columns(group, 'detflux')
    merged_detflux_err = concatenate_list_columns(group, 'detflux_err')

    return pd.Series({
        'merged_time': merged_time,
        'merged_detflux': merged_detflux,
        'merged_detflux_err': merged_detflux_err,
        'total_n_points': total_n_points
    })

KCC_processed = KCC.copy()

# Aggregate data by KIC
result_df = KCC_processed.groupby('KIC').apply(aggregate_kic_data)

```

```python
def plotLightCurve(df, kic, with_error=True):
    """
    Plot the light curve for a given KIC from a DataFrame.
    """
    row = df[df["KIC"] == kic]
    
    if row.empty:
        raise ValueError(f"KIC {kic} not found in DataFrame.")
    
    row = row.iloc[0]
    
    t = row["time"]
    f = row["flux"]
    e = row.get("flux_err", None)

    # Plot
    plt.figure(figsize=(20, 5))
    
    if with_error and e is not None:
        plt.errorbar(t, f, yerr=e, fmt='-', linewidth=1)
    else:
        plt.plot(t, f, linewidth=1)
    
    plt.xlabel("Time")
    plt.ylabel(r"Normalized Flux (e$^{-}$ s$^{-1}$)")
    plt.title(f"KIC {kic} Light Curve")
    plt.show()
```

```python
def sampleRandomKIC(df):
    """Samples a random KIC value from the DataFrame index."""
    return int(np.random.choice(df.index))
```
